## Task A

You will analyze physiological measurements connecting carbohydrate
intake ($x$, in grams) with blood glucose response ($y$, in mg/dL).

Our goal is to **model** $y$ as a function of $x$:

### Task A.1

Below is the measurement dataset recorded for two patients. Our goal is
to model $y$ as a function of $x$.

| Observation | Carbohydrate Intake $x$ (g) | Blood Glucose Response $y$ (mg/dL) |
|:--------------------|:------------------------:|:------------------------:|
| **Patient 1** | $20$ | $40$ |
| **Patient 2** | $50$ | $100$ |

1.  **Visualize the Data**: Plot the two patient measurements on a
    scatter plot ($x$ vs. $y$), using matplotlib. Label axes and add a
    title to the plot.

2.  **Model & Formulate Equations**: On paper, write down a simple
    linear model equation $$y_i = \theta_0 + \theta_1 x_i$$ for each
    patient. Enter the observed data values to set up a **system of
    linear equations**.

3.  **Manual Analysis**:

    -   How many observations are in this dataset? How many unknown
        parameters are we solving for?
    -   Solve the system of linear equations manually (by inspection or
        by hand on paper) to find the unknown parameters.

4.  **Matrix Formulation & Matrix Inversion**:

    -   Write the system of linear equations in matrix form
        $$\mathbf{y} = \mathbf{X}\boldsymbol{\theta}$$. Specify all
        values in $\mathbf{X}$
    -   Determine the dimensions and shapes of the target vector
        $\mathbf{y}$, the matrix $\mathbf{X}$, and the parameter vector
        $\boldsymbol{\theta}$.
    -   How would you write down the solution for $\boldsymbol{\theta}$
        if you knew the inverse $\mathbf{X}^{-1}$?

5.  **Numerical Python Implementation**:

    -   Define vectors and matrices in Python and use NumPy to compute
        $\boldsymbol{\theta}$ numerically.
    -   Compare your numerical result with your manual solution.
    -   plot the predicted line $y = \theta_0 + \theta_1 x$ along with
        the original data points.
    -   *Hint*: Useful NumPy functions include `np.column_stack()`,
        `np.ones()`, `np.linalg.inv()`, and `np.linalg.solve()`. For
        matrix definitions and calculation methods, refer to the
        **[Maths to NumPy](../../references/math_to_numpy.qmd)**
        dictionary.
